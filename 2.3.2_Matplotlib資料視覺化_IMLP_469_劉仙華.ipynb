## 4 如果畫個參數式圖形

其實參數式圖形基本上也是一堆 x, y 的座標, 只是 x(t), y(t) 都是 t 的函數, 我們可以想成 t 是時間。

假設我們的

$$ -2\pi \leq t \leq 2\pi$$

要畫個半徑是 r 的圓, x(t), y(t) 就是這樣:

* $x(t) = r\cos(t)$
* $y(t) = r\sin(t)$

我們來試試事情是不是如我們想的這樣...

```python
%matplotlib inline
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

r = 3

t = np.linspace(-2*np.pi, 2*np.pi, 200 )

x = r*np.cos(t)
y = r*np.sin(t)

plt.plot(x, y, c="g", lw=5)
```

真的成功了... 等等, `matplotlib` 怎麼畫成橢圓呢? 我們來調整一下。

```python
%matplotlib inline
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

r = 3

t = np.linspace(-2*np.pi, 2*np.pi, 200 )

x = r*np.cos(t)
y = r*np.sin(t)

ax = plt.gca()          # 把 x 軸 和 y 軸 長寬比 調整相同, 才能形成一個圓形
ax.set_aspect("equal")

plt.plot(x, y, c="g", lw=5)
```

再來我們可以想一下, 如果 r 不是故定的呢? 也就是它會變長變短, 是不是會畫出很多特別的圖呢? 其實很多特別的曲線不過就是找個試合的 r(t), 我們來隨意試試看。

```python
%matplotlib inline
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

t = np.linspace(-2*np.pi, 2*np.pi, 200 )
r = np.sin(5*t)

x = r*np.cos(t)
y = r*np.sin(t)

ax = plt.gca()          # plt.gca() 代表 "Get Current Axes"（取得目前的座標軸)
ax.set_aspect("equal")  # 把 x 軸 和 y 軸 長寬比 調整相同

plt.plot(x, y, c="b", lw=3)
```

是不是很有意思呢? 你也可以自己試試不同的變化。

## 5 `subplot` 畫多個圖

我們每次畫圖的時候, `matplotlib` 就弄 1 個 figure 畫圖區出來, 裡面可以有很多子圖, 在 figure 裡叫 axes。目前我們都只有 1 個 figure 內含 1 張圖, 所以都不用設, 現在我想畫 4 張圖時。我們就要先想好「陣式」。

比如說 2x2 這樣排列的 4 張圖。

```python
%matplotlib inline
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

x = np.linspace(0, 10, 100)

plt.subplot(221)   # 2x2 共4個格位, 1 即是第1個格位
plt.plot(x, np.sin(x), c="#e63946", lw=3)

plt.subplot(222)   # 2x2 共4個格位, 2 即是第2個格位
plt.plot(x, np.sin(x), c="#7fb069", lw=3)

plt.subplot(223)   # 2x2 共4個格位, 3 即是第3個格位
plt.scatter(x, np.random.rand(100), c="#daa73e", s=50, alpha=0.5)

# 在 plt.scatter() 中，s 代表 "Size"（尺寸），用來控制散佈圖中點的大小
# 在 plt.scatter() 中，alpha 代表「透明度」（Opacity）

plt.subplot(224)   # 2x2 共4個格位, 4 即是第4個格位
plt.bar(range(10), np.random.randint(1, 30, 10), fc='#e59344')

# np.random.randint 三個參數的意思
# 1 （下限）：產生的隨機數包含 1
# 30（上限）：產生的隨機數不包含 30（最高只到 29）
# 10（數量）：總共要產生 10 個隨機整數, 數值必須要和 range() 內容數值 相同
# 在 plt.bar() 中，fc 代表 "Face Color"（面色/填滿顏色），用來控制長條圖柱子內部的填滿顏色。
```

所謂 'marker' 就是原來的點, 我們特別標示出來, 為了不要讓圖形太複雜, 我們新換個點比較少的例子。

#### marker 可以設的參數

| 參數 | 說明 |
|---|---|
| `marker` | marker 的風格 |
| `markeredgecolor` (`mec`) | 邊線顏色 |
| `markeredgewidth` (`mew`) | 邊線寬度 |
| `markerfacecolor` (`mfc`) | marker 的顏色 |
| `markerfacecoloralt` (`mfcalt`) | marker 替換色 |
| `markersize` (`ms`) | marker 大小 |
| `markevery` | 隔多少畫一個 marker |

```python
%matplotlib inline
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

x = range(20)
y = np.random.randn(20)

plt.plot(x, y, c="#62a87c", lw=5, marker='o', mfc='#ff9b71', mec='#313b72', ms=20)
```

比較奇特的參數是 `markevery`, 這是說我們每多少筆資料要畫一個 marker。為了示範, 我們拿之前的例子來看看。

```python
%matplotlib inline
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

x = np.linspace(-1, 10, 500)   # -1 到 10, 給 500 個點
y = np.sin(5*x)/(1+x**2)
plt.plot(x, y, lw=3, marker='o', ms=10)
```

後面的完完全全連在一起了, 我們現在隔 10 個畫一個 marker 試試。

```python
%matplotlib inline
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

x = np.linspace(-1, 10, 500)   # -1 到 10, 給 500 個點
y = np.sin(5*x)/(1+x**2)
plt.plot(x, y, lw=3, marker='o', ms=10, markevery=10)    # markevery() : 設定每幾多個點才顯示一個點
```

## 10 取得現在工作中 axes*

我們有時要設 axes 的背景啦等等的資訊。這時就要取得現在工作中的 axes。這一般有兩種方式, 第一種是設 `subplot` 時可以取得:

    fig, ax = plt.subplot()
    
另一種是用 `gca` 函數:

    ax = plt.gca()

```python
%matplotlib inline
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

ax = plt.gca()                  # plt.gca() 代表 "Get Current Axes"（取得目前的座標軸)
ax.set_facecolor('#69b8bb')

ax.set_xlim(-6, 6)
ax.set_ylim(-1.2, 1.2)

plt.plot(x, y, lw=5, c='white')
```

#### 移動 x, y 座標軸

```python
%matplotlib inline
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

ax = plt.gca()                  # plt.gca() 代表 "Get Current Axes"（取得目前的座標軸)
ax.set_facecolor('#69b8bb')

ax.set_xlim(-6, 6)
ax.set_ylim(-1.2, 1.2)

ax.spines['bottom'].set_position(('data', 0))
ax.spines['left'].set_position(('data', 0))

plt.plot(x, y, lw=5, color='white')
```
