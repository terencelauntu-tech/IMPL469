```python
#匯入在sklearn.linear_model套件裡面的LinearRegression模型
#匯入在sklearn.model_selection套件裡面的train_test_split模組
#切割數據集(訓練資料80%、測試資料20%，設定random_state=20)
#畫出訓練資料集的matplotlib圖形m
```

```python
print(x.shape)
```

```python
XR = x.reshape(-1, 1)   # 1D --> 2D
print(XR.shape)
```

```python
#將原始資料集隨機切分為 訓練集（Training Set）80% 與 測試集（Testing Set) 20%
x_train, x_test, y_train, y_test = train_test_split(x, y, test_size=0.2, random_state=20) 
print(x_train)
```

```python
y
```

```python
plt.scatter(x_train, y_train)
```

```python
regr = LinearRegression()
X_train = x_train.reshape(-1,1) 
regr.fit(X_train,y_train)
print('Coefficients: ', regr.coef_)         # Coefficients : Weight（權重） 或 Slope（斜率）
print('interception: ', regr.intercept_)    # interception : 截距  ( Bias（偏置）)
```

```python
Y_train = regr.predict(X_train)
print(Y_train)
```

```python
mse = np.sum((Y_train-y_train)**2) / len(y_train)
print(mse)     # mse : Mean Squared Error 均方誤差
```

```python
X_test = x_test.reshape(-1,1)
Y_test = regr.predict(X_test)

print(Y_test)
```

```python
mse = np.sum((Y_test-y_test)**2) / len(y_test)
print(mse)    # Test data mse higher then train data mse   (mse : Mean Squared Error 均方誤差 )
```
